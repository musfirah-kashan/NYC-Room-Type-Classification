```python
sns.countplot(x='room_type',data=df)
plt.tight_layout()
plt.savefig('graphs/room_type_distribution.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
numeric_cols=['price','minimum_nights','number_of_reviews','reviews_per_month','calculated_host_listings_count','availability_365']
df[numeric_cols].hist(bins=30,figsize=(12,8))
plt.tight_layout()
plt.savefig('graphs/airbnb_numeric_plots.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
sns.countplot(data=df,x='neighbourhood_group')
plt.tight_layout()
plt.savefig('graphs/neighbourhood_group_plot.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
sns.boxplot(x='room_type',y='price',data=df)
plt.tight_layout()
plt.savefig('graphs/room_type_plot.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
corr=df[numeric_cols+['latitude','longitude']].corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.savefig('graphs/correlation_heatmap.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
sns.countplot(x='neighbourhood_group', hue='room_type', data=df)
plt.title('Room Type Distribution Across Neighbourhoods')
plt.savefig('graphs/neighbourhood_room_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
sns.scatterplot(x='longitude', y='latitude', data=df, hue='room_type')
plt.title('Geographic Distribution of Listings by Room Type')
plt.savefig('graphs/nyc_listings_scatterplot.png', dpi=300, bbox_inches='tight')
plt.show()
```
